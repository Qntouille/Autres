```python
collection_base = collection.copy()

cartes_cibles = extension_potentielles_reverse_pokeballs_masterballs.keys()

s = 0

liste_nb_iterations = []

for j in range(1000) :
    collection = collection_base.copy()
    masterset = False
    i = 0

    # On en ouvre 5000 d'un coup dès le départ, ça permet de faire moins de tests
    for k in range(5000) :
        i += 1
        ouverture_booster_aleatoire()

    while masterset == False : 
        pokeballs = False
        masterballs = False
        reverse = False
        serie = False
        i += 1
        
        ouverture_booster_aleatoire()

        # Met à jour les sous-ensembles à chaque itération
        subset_reverse = collection.loc[collection.index.isin(cartes_cibles), 'Reverse']
        subset_pokeball = collection.loc[collection.index.isin(cartes_cibles), 'Pokeball']
        subset_masterball = collection.loc[collection.index.isin(cartes_cibles), 'Masterball']

        if (collection['Nombre obtenus'] != 0).all() :
            serie = True
        if (subset_reverse != 0).all() :
            reverse = True
        if (subset_pokeball != 0).all() :
            pokeballs = True
        if (subset_masterball != 0).all() : 
            masterballs = True
        
        if [serie, reverse, pokeballs, masterballs] == [True, True, True, True] :
            masterset = True
            print(f'{j+1} mastersets faits, {i} boosters ouverts pour ce dernier')

    liste_nb_iterations.append(i)
    s += i

print(s/1000)
```
